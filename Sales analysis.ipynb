```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Phone", "Mouse",
        "Laptop", "Keyboard", "Phone", "Mouse", "Laptop",
        "Phone", "Keyboard", "Mouse", "Laptop", "Phone"
    ],
    "City": [
        "Sukkur", "Karachi", "Sukkur", "Karachi", "Hyderabad",
        "Karachi", "Hyderabad", "Sukkur", "Sukkur", "Hyderabad",
        "Karachi", "Karachi", "Hyderabad", "Sukkur", "Hyderabad"
    ],
    "Quantity": [
        2, 5, 3, 2, 4, 1, 2, 3, 6, 1, 1, 4, 3, 1, 2
    ],
    "Unit_Price": [
        120000, 1500, 2500, 80000, 1500,
        120000, 2500, 80000, 1500, 120000,
        80000, 2500, 1500, 120000, 80000
    ]
}

df = pd.DataFrame(data)

df.to_csv("sales.csv", index=False)

print("SALES DATA ANALYSIS PROJECT")
print("=" * 40)

print("\n1. First Five Records:")
print(df.head())

print("\n2. Dataset Information:")
print("Total Orders:", len(df))
print("Total Columns:", len(df.columns))
print("Columns:", df.columns.tolist())

print("\n3. Missing Values:")
print(df.isnull().sum())

print("\n4. Duplicate Rows:", df.duplicated().sum())

df["Revenue"] = df["Quantity"] * df["Unit_Price"]

print("\n5. Complete Sales Data:")
print(df.to_string(index=False))

print("\n6. Total Sales Revenue:")
print("Rs.", f"{df['Revenue'].sum():,.2f}")

print("\n7. Total Units Sold:")
print(df["Quantity"].sum())

print("\n8. Average Revenue per Order:")
print("Rs.", f"{df['Revenue'].mean():,.2f}")

product_sales = df.groupby("Product")["Revenue"].sum()
city_sales = df.groupby("City")["Revenue"].sum()

print("\n9. Revenue by Product:")
print(product_sales.sort_values(ascending=False))

print("\n10. Revenue by City:")
print(city_sales.sort_values(ascending=False))

print("\n11. Highest Revenue Product:")
print(product_sales.idxmax())

print("\n12. Highest Revenue City:")
print(city_sales.idxmax())

plt.figure(figsize=(8, 5))
product_sales.sort_values(ascending=False).plot(
    kind="bar",
    title="Revenue by Product",
    xlabel="Product",
    ylabel="Revenue (Rs.)"
)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("product_sales.png")
plt.show()

plt.figure(figsize=(8, 5))
city_sales.sort_values(ascending=False).plot(
    kind="bar",
    title="Revenue by City",
    xlabel="City",
    ylabel="Revenue (Rs.)"
)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("city_sales.png")
plt.show()

df.to_csv("sales_with_revenue.csv", index=False)

print("\nProject Completed Successfully!")
print("Files saved: sales.csv, sales_with_revenue.csv")
print("Charts saved: product_sales.png, city_sales.png")
```
